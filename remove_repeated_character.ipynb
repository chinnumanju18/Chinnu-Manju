```python
def remove_repeated_chars_dict(words):
    def remove_repeats(word):
        seen = set()
        result = []
        for char in word:
            if char not in seen:
                seen.add(char)
                result.append(char)
        return ''.join(result)
    return {word: remove_repeats(word) for word in words}
words = ["apple", "banana", "cherry", "date"]
filtered_words_dict = remove_repeated_chars_dict(words)
print(filtered_words_dict)
```
